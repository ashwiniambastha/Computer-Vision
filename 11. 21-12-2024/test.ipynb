```python
import torch
import torch.nn as nn
```

```python
class LinearR(nn.Module):
    def __init__(self):
        super(LinearR,self).__init__()
        self.linear=nn.Linear(in_features=1,out_features=1)
    
    def forward(self,x):
        out=self.linear(x)
        return out
```

```python
x=torch.rand(size=(4,1))
y=torch.rand(size=(4,1))
```

```python
model=LinearR()
```

```python
cre=nn.MSELoss()
optimizer=torch.optim.SGD(model.parameters(),lr=0.01)
```

```python
e=1000
for epoch in range(e):
    
    y_pred=model(x)
    loss=cre(y,y_pred)
    
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
    
    if epoch%100==0:
        print(f"For Epoch:{epoch} , the loss is {loss}\n")
```
