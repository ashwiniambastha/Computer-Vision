## Sequential Way


```python
import torch
import torch.nn as nn
```

```python
seq=nn.Sequential(
    nn.Linear(10,5),
    nn.ReLU(),
    nn.Linear(5,15),
    nn.ReLU(),
    nn.Linear(15,2),
)
```

```python
x=torch.rand(20,10)
```

```python
neural=seq(x)
```

```python
print(neural)
```
