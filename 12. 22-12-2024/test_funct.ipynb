```python
import torch
import torch.nn as nn
```

```python
class LinearR(nn.Module):
    def __init__(self):
        super(LinearR,self).__init__()
        self.fl1=nn.Linear(2,4)
        self.fl2=nn.Linear(4,2)
        self.fl3=nn.Linear(2,3)
        self.relu=nn.ReLU()
    
    def forward(self,x):
        x=self.relu(self.fl1(x))
        x=self.relu(self.fl2(x))
        x=self.fl3(x)
        
        return x
```

```python
x=torch.randn(5,2)
```

```python
model=LinearR()
```

```python
output=model(x)
```
