```python
import torch
import torch.nn as nn
```

```python
class Li(nn.Module):
    def __init__(self):
        super(Li,self).__init__()
        self.fl1=nn.Linear(5,20)
        self.fl2=nn.Linear(20,32)
        self.fl3=nn.Linear(32,3)
        self.relu=nn.ReLU
    
    def forward(self,x):
        x=self.relu(self.fl1(x))
        x=self.relu(self.fl2(x))
        x=self.fl3(x)
        
        return x
        
```
